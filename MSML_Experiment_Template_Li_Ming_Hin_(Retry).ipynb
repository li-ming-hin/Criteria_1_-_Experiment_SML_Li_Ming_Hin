# **5. Data Preprocessing**

At this stage, data preprocessing is a crucial step to ensure data quality before use in a machine learning model.

If you're using text data, the raw data often contains blanks, duplicates, or inconsistent value ranges, which can impact model performance. Therefore, this process aims to clean and prepare the data for optimal analysis.

The following steps can be performed, but are not limited to:
1. Removing or Handling Missing Values
2. Removing Duplicate Data
3. Normalizing or Standardizing Features
4. Detecting and Handling Outliers
5. Encoding Categorical Data
6. Binning (Data Grouping)

Simply adapt this to the characteristics of the data you're using, especially when using unstructured data.

```python
# 5.1 Handling Missing Values

df.isnull().sum()
```

```python
# If missing values exist → fill or drop
df = df.dropna()
```

```python
# 5.2 Removing Duplicate Data

df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
# 5.3 Handling Outliers (IQR Method)

import numpy as np

def remove_outliers(df, cols):
    for col in cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        df = df[(df[col] >= lower) & (df[col] <= upper)]
    return df

numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns
df = remove_outliers(df, numerical_cols)
```

```python
# 5.4 Encoding Categorical Data

categorical_cols = df.select_dtypes(include=['object']).columns
categorical_cols
```

```python
df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
```

```python
# 5.5 Feature Scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X = df.drop("target", axis=1)
y = df["target"]

X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame
X = pd.DataFrame(X_scaled, columns=X.columns)
```

```python
# 5.6 Final Clean Dataset

df_clean = pd.concat([X, y.reset_index(drop=True)], axis=1)
df_clean.head()
```

```python
# 5.7 Save Preprocessed Dataset

import os

output_dir = "preprocessing_dataset"
os.makedirs(output_dir, exist_ok=True)

df_clean.to_csv(f"{output_dir}/heart_clean.csv", index=False)

print("Preprocessed dataset saved successfully!")
```
